```r

find_median <- function(x) {
  x <- sort(x)
  n <- length(x)

  if (n %% 2 == 1) {
    median <- x[(n + 1) / 2]
  } else {
    median <- (x[n / 2] + x[n / 2 + 1]) / 2
  }

  return(median)
}

x <- c(3, 7, 2, 9, 5)
find_median(x)
```

```r
x <- c(8, 9, 5, 12, 20, 15)
find_median(x)
```

```r
mean_extend <- function(x) {
  while (length(x) < 10) {
    mean_x <- sum(x) / length(x)
    x <- c(x, mean_x)

    print(x)
    print(mean_x)
  }

  return(x)
}

x <- c(3, 4, 5, 6)
mean_extend(x)
```

```r
x <- c(9, 10, 6, 12, 8, 8)
mean_extend(x)
```

```r
hitung_genap <- function(x) {
  jumlah <- 0

  for (i in x) {
    if (i %% 2 == 0) {
      jumlah <- jumlah + 1
    }
  }

  return(jumlah)
}

x <- c(3, 4, 7, 10, 12, 15)
hitung_genap(x)
```

```r
x <- c(8, 8, 10, 9, 5, 100, 3)
hitung_genap(x)
```

```r
hitung_genap_matriks <- function(m) {
  jumlah <- 0

  for (i in 1:nrow(m)) {
    for (j in 1:ncol(m)) {
      if (m[i, j] %% 2 == 0) {
        jumlah <- jumlah + m[i, j]
      }
    }
  }

  return(jumlah)
}

m <- matrix(1:12, nrow = 3, ncol = 4)

m

hitung_genap_matriks(m)
```

```r
m <- matrix(2:22, nrow = 4, ncol = 5)
m
hitung_genap_matriks(m)
```
